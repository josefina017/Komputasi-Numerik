```
# SPL 3 VARIABEL - ELIMINASI GAUSS
# Nama: Josefina Baqita De Assuncao | NIM: 2511500014

import numpy as np

# Contoh dari Algoritma No.5 saya:
# 2x + 1y + 1z = 7
# 1x + 3y + 2z = 13
# 1x + 0y + 2z = 7

A = np.array([[2., 1., 1.],
              [1., 3., 2.],
              [1., 0., 2.]])
b = np.array([7., 13., 7.])

print("Matriks A:")
print(A)
print("Nilai b:", b)
```

```
# Gabung A dan b jadi matriks augmented
n = 3
M = np.hstack([A, b.reshape(-1, 1)])
print("Matriks Augmented Awal:")
print(M)
```

```
# ELIMINASI: buat 0 di bawah diagonal
for k in range(n-1):
    for i in range(k+1, n):
        faktor = M[i, k] / M[k, k]
        M[i] = M[i] - faktor * M[k]
        print(f"\nSetelah eliminasi baris {i+1}:")
        print(M)

print("\nMatriks sudah bentuk segitiga atas:")
print(M)
```

```
# SUBSTITUSI BALIK: cari z dulu, lalu y, lalu x
x = np.zeros(n)
for i in range(n-1, -1, -1):
    jumlah = sum(M[i, j] * x[j] for j in range(i+1, n))
    x[i] = (M[i, n] - jumlah) / M[i, i]

print(f"Hasilnya: x = {x[0]}, y = {x[1]}, z = {x[2]}")
```

```
# CEK ULANG: masukkan x,y,z ke persamaan awal
cek = A @ x
print("A dikali x =", cek)
print("Harusnya sama dengan b =", b)
if np.allclose(cek, b):
    print("BENAR! Jawaban sudah tepat.")
else:
    print("Salah, cek lagi.")
```
