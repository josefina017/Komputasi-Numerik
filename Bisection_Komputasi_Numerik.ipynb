```python
# Nama : Josefina Baqita De Assuncao
# NIM : 202511420096
# Topik : Bisection - Cari akar x^2 - 4 = 0

# Ini fungsi yang mau kita cari akarnya
def f(x):
    return x**2 - 4  # f(x) = x kuadrat dikurang 4

# Batas awal saya
a = 1  # Batas kiri, f(1) = -3 (negatif)
b = 3  # Batas kanan, f(3) = 5 (positif)
print("f(1) =", f(a))
print("f(3) =", f(b))
print("Karena tandanya beda, akar ada di antara 1 dan 3")
```

```python
# Setting seberapa teliti saya mau
toleransi = 0.001  # Kalau jarak a dan b sudah < 0.001, berhenti
maks_iterasi = 20  # Maksimal 20 kali bagi dua, biar tidak jalan terus
print("Toleransi saya:", toleransi)
print("Maksimal iterasi:", maks_iterasi)
```

```python
# Proses utama Bisection, bagi dua terus
for i in range(1, maks_iterasi + 1):  # Ulangi dari iterasi 1 sampai 20
    c = (a + b) / 2  # Hitung titik tengah
    print(f"Iterasi {i}: a={a:.4f}, b={b:.4f}, c={c:.4f}, f(c)={f(c):.4f}")

    if f(c) == 0:  # Kalau f(c) pas 0, berarti ketemu persis
        break
    if f(a) * f(c) < 0:  # Kalau tanda f(a) dan f(c) beda, akar di kiri
        b = c  # Pindahkan batas kanan ke tengah
    else:  # Kalau tidak, berarti akar di kanan
        a = c  # Pindahkan batas kiri ke tengah

    if (b - a) < toleransi:  # Kalau jarak sudah kecil banget
        break  # Berhenti, sudah cukup teliti
```

```python
# Hasil akhirnya adalah titik tengah terakhir
akar = (a + b) / 2
print("Akar yang ketemu:", akar)
print("Jumlah iterasi:", i)
```

```python
# Cek dengan masukkan akar ke fungsi awal
hasil_cek = f(akar)
print("f(akar) =", hasil_cek)
if abs(hasil_cek) < 0.01:  # Kalau hasilnya mendekati 0
    print("BENAR! Jawaban sudah tepat.")
else:
    print("Belum tepat, coba cek lagi.")
```
