```
# CELL 1: MEMBUAT MATRIKS DARI SOAL
# Import library numpy untuk perhitungan matriks, disingkat jadi 'np'
import numpy as np

# Matriks A = koefisien dari x, y, z
# Baris 1: 2x + 1y + 1z -> [2, 1, 1]
# Baris 2: 1x + 3y + 2z -> [1, 3, 2]
# Baris 3: 1x + 0y + 2z -> [1, 0, 2] (0 karena tidak ada y)
A = np.array([[2., 1., 1.],
              [1., 3., 2.],
              [1., 0., 2.]])

# Vektor b = hasil di ruas kanan sama dengan
# Persamaan 1 = 7, Persamaan 2 = 13, Persamaan 3 = 7
b = np.array([7., 13., 7.])

# Menampilkan matriks A dan vektor b untuk memastikan input benar
print("Matriks A:")
print(A)
print("Nilai b:", b)
```

```
# CELL 2: MEMBUAT MATRIKS AUGMENTED
# n = jumlah variabel, ada 3 yaitu x, y, z
n = 3

# Menggabungkan matriks A dan vektor b menjadi satu matriks
# np.hstack = menempel secara horizontal (ke kanan)
# b.reshape(-1,1) = mengubah b dari [7,13,7] menjadi kolom tegak
# Hasilnya matriks 3x4, kolom terakhir adalah nilai b
M = np.hstack([A, b.reshape(-1, 1)])
print("Matriks Augmented Awal:")
print(M)
```

```
# CELL 3: ELIMINASI GAUSS - MEMBUAT SEGITIGA ATAS
# Tujuan: membuat semua angka di bawah diagonal utama menjadi 0
# Loop luar (k): mengerjakan kolom demi kolom, dari kolom 0 sampai 1
for k in range(n-1):
    # Loop dalam (i): mengerjakan baris di bawah baris k
    for i in range(k+1, n):
        # Menghitung faktor pengali
        # Rumus: angka yang mau di-nol-kan dibagi angka diagonal
        faktor = M[i, k] / M[k, k]
        # Operasi Baris Elementer: Baris i dikurangi (faktor x Baris k)
        # Hasilnya M[i,k] akan menjadi 0
        M[i] = M[i] - faktor * M[k]
        print(f"\nSetelah eliminasi baris {i+1}:")
        print(M)

# Menampilkan hasil akhir setelah semua eliminasi selesai
print("\nMatriks sudah bentuk segitiga atas:")
print(M)
```

```
# CELL 4: SUBSTITUSI BALIK - MENCARI NILAI x, y, z
# Membuat tempat kosong untuk menyimpan hasil x, y, z, awalnya [0,0,0]
x = np.zeros(n)
# Loop dari baris paling bawah ke atas (2, 1, 0)
# Karena kita mulai dari bawah, z akan ketemu dulu, lalu y, lalu x
for i in range(n-1, -1, -1):
    # Menghitung jumlah dari variabel yang sudah diketahui
    # Contoh di baris 1: kalau sudah tau y dan z, kita hitung (koef*y + koef*z)
    jumlah = sum(M[i, j] * x[j] for j in range(i+1, n))
    # Rumus utama: x[i] = (nilai kanan - jumlah) / koefisien diagonal
    x[i] = (M[i, n] - jumlah) / M[i, i]

# Menampilkan hasil akhir
print(f"Hasilnya: x = {x[0]}, y = {x[1]}, z = {x[2]}")
```

```
# CELL 5: VERIFIKASI / CEK ULANG JAWABAN
# Mengalikan kembali Matriks A awal dengan hasil x yang didapat
# Operator @ artinya perkalian matriks, bukan perkalian biasa
cek = A @ x
print("A dikali x =", cek)
print("Harusnya sama dengan b =", b)
# np.allclose = mengecek apakah dua nilai hampir sama / sama persis
# Jika sama, berarti jawaban kita BENAR, jika tidak sama berarti ada kesalahan hitung
if np.allclose(cek, b):
    print("BENAR! Jawaban sudah tepat.")
else:
    print("Salah, cek lagi.")
```
